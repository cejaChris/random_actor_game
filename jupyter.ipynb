```python
import pandas as pd
from bs4 import BeautifulSoup as bs
import requests
```

```python
# trick websites into thinking the program is a browser
header = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36",
    "Accept": "text/html,application/xhtml+xml,application/xml;q=0.9,image/webp,*/*;q=0.8",
    "Accept-Language": "en-US,en;q=0.5",
    "Connection": "keep-alive",
}
url = requests.get('https://www.the-numbers.com/box-office-star-records/worldwide/lifetime-acting/top-grossing-leading-stars', headers=header)
soup = bs(url.text, 'html.parser')
soup
```

```python
table = soup.find('tbody')
table
```

```python
tr = table.find_all('tr')
tr
```

```python
columns = []
for row in tr:
    row_data = row.find_all('td')
    individual_row_data = [data.text.strip() for data in row_data]
    columns.append(individual_row_data)

columns
```

```python
# finding the headers
thead = soup.find('thead')
thead
```

```python
headers = []
th = thead.find_all('th')
for head in th:
    headers.append(head.text.strip())
headers
```

```python
# making the df

df_dict = {}

for header in headers:
    df_dict[header] = []

for column in columns:
    for data, header in zip(column, headers):
        df_dict[header].append(data)

df = pd.DataFrame(df_dict)
df
```
